속도/품질 프리셋:

- `--preset debug`: 코드 테스트용
- `--preset fast`: 초반 추천, 훨씬 빠른 반복
- `--preset balanced`: 품질/속도 균형
- `--preset strong`: A100 이상 긴 학습용
- GPU에서는 `mcts-batch-size`가 8~32 정도면 모델 호출을 묶어서 더 빠르게 돌 수 있습니다.
- self-play는 초반 temperature를 높게, 중후반 temperature를 낮게 쓰도록 스케줄링됩니다.
- CUDA 모델 self-play에서는 GPU 메모리 보호를 위해 self-play workers가 자동으로 1로 내려갈 수 있습니다.

`eval.json`의 `win_rate`, `elo_diff`, `wins/draws/losses`를 보면 개선 여부가 보입니다.
