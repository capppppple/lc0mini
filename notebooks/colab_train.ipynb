학습량은 주로 `--games`, `--simulations`, `--epochs`, `--batch-size`로 조절합니다.

- 빠른 테스트: games 5, simulations 8, epochs 1

- eval.json의 win_rate, elo_diff, wins/draws/losses를 보면 개선 여부가 보입니다.
